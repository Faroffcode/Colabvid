```python
# Install dependencies (Colab / Python 3.13 compatible)
import sys, subprocess, shutil
def pip_install(*packages):
    subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', '--upgrade', *packages], check=True)
if shutil.which('ffmpeg') is None:
    subprocess.run(['apt-get','update','-qq'], check=True)
    subprocess.run(['apt-get','install','-y','-qq','ffmpeg'], check=True)
pip_install('pip', 'setuptools', 'wheel')
pip_install('gradio>=6,<7', 'requests', 'pyscenedetect[opencv]', 'telethon')
# Whisper is optional; install it separately so a Whisper dependency failure does not hide the main app.
try:
    pip_install('openai-whisper')
    print('✅ Whisper installed')
except subprocess.CalledProcessError:
    print('⚠️ Whisper could not be installed. Colabvid will still run; disable Local Whisper subtitles for now.')
print('✅ Core dependencies installed')
```

```python
# Launch Colabvid
import sys
sys.path.insert(0, '/content/Colabvid')
import colabvid
app = colabvid.ui()
app.queue().launch(share=True, show_error=True, max_file_size='20gb')
```
