```python
# Install dependencies one package at a time so the failing package is visible
import sys, subprocess, shutil
def install_pkg(pkg):
    print(f'Installing {pkg}...')
    r=subprocess.run([sys.executable,'-m','pip','install','-q','--disable-pip-version-check',pkg])
    if r.returncode != 0:
        raise RuntimeError(f'Failed to install: {pkg}')
    print(f'✅ {pkg}')
if shutil.which('ffmpeg') is None:
    subprocess.run(['apt-get','update','-qq'],check=True)
    subprocess.run(['apt-get','install','-y','-qq','ffmpeg'],check=True)
install_pkg('gradio>=6,<7')
install_pkg('requests')
install_pkg('pyscenedetect[opencv]')
install_pkg('telethon')
# Whisper is optional. The main app can run without it.
try:
    install_pkg('openai-whisper')
    WHISPER_OK=True
except Exception as e:
    WHISPER_OK=False
    print('⚠️ Whisper install failed; disable Local Whisper subtitles for now.')
print('✅ Core Colabvid dependencies installed')
```

```python
# Launch Colabvid
import sys
sys.path.insert(0, '/content/Colabvid')
import colabvid
app = colabvid.ui()
app.queue().launch(share=True, show_error=True, max_file_size='20gb')
```
