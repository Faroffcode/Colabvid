```python
# Install dependencies (PySceneDetect is optional; fallback segmentation is built in)
import sys, subprocess, shutil
def install_pkg(pkg, optional=False):
    print(f'Installing {pkg}...')
    r=subprocess.run([sys.executable,'-m','pip','install','--disable-pip-version-check',pkg])
    if r.returncode != 0:
        if optional: print(f'⚠️ Optional package unavailable: {pkg}')
        else: raise RuntimeError(f'Failed to install: {pkg}')
    else: print(f'✅ {pkg}')
if shutil.which('ffmpeg') is None:
    subprocess.run(['apt-get','update','-qq'],check=True)
    subprocess.run(['apt-get','install','-y','-qq','ffmpeg'],check=True)
install_pkg('gradio>=6,<7')
install_pkg('requests')
install_pkg('telethon')
install_pkg('pyscenedetect', optional=True)
try:
    install_pkg('openai-whisper', optional=True)
except Exception: pass
print('✅ Colabvid core dependencies installed')
```

```python
# Launch Colabvid
import sys
sys.path.insert(0, '/content/Colabvid')
import colabvid
app = colabvid.ui()
app.queue().launch(share=True, show_error=True, max_file_size='20gb')
```
